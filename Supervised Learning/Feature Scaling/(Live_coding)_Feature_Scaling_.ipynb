# Standardization

```python
#define the variables I wanna scale
cols_to_scale = ["StudyHoursPerWeek", "AttendanceRate", "GPA"]

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[cols_to_scale])
scaled_df = pd.DataFrame(scaled_data, columns=cols_to_scale)

scaled_df.head()
```

## Validate Our Calculations (for Standardization)

Mean should be 0
and variance should be 1

However, as you can see in the next calculations we see the mean has a negative value and a lot of digits next to it. It happens due to two main reasons (Floating-point precision limits) and (Degrees of Freedom (ddof))

1. **Floating-point precision limits**: Computers represent real numbers using binary floating points. Calculations like subtracting the mean can leave extremely tiny residual decimals (e.g., $10^{-16}$$10^{-16}$). This is practically equivalent to 0.

2. **Degrees of Freedom (ddof)**: By default, scikit-learn's StandardScaler standardizes your data using Population Variance (divided by $N$$N$). However, Pandas' .var() function calculates Sample Variance by default (divided by $N - 1$$N - 1$, which is ddof=1).

*To ensure we get the same results as we did in the excel file, kindly check the 2nd code cell*

```python
# for standardization
print(f"'StudyHoursPerWeek' has Exact Mean: {scaled_df["StudyHoursPerWeek"].mean()}, Exact Variance: {scaled_df["StudyHoursPerWeek"].var()}")
print(f"'AttendanceRate' has Exact Mean: {scaled_df["AttendanceRate"].mean()}, Exact Variance: {scaled_df["AttendanceRate"].var()}")
print(f"'GPA' has Exact Mean: {scaled_df["GPA"].mean()}, Exact Variance: {scaled_df["GPA"].var()}")
```

```python
# enhancing the numbers of standardization
print(f"'StudyHoursPerWeek' has Exact Mean: {abs(scaled_df["StudyHoursPerWeek"].mean().round())}, Exact Variance: {round(scaled_df["StudyHoursPerWeek"].var())}")
print(f"'AttendanceRate' has Exact Mean: {abs(scaled_df["AttendanceRate"].mean().round())}, Exact Variance: {round(scaled_df["AttendanceRate"].var())}")
print(f"'GPA' has Exact Mean: {abs(scaled_df["GPA"].mean().round())}, Exact Variance: {round(scaled_df["GPA"].var())}")
```
